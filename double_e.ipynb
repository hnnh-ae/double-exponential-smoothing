DOUBLE EXPONENTIAL SMOOTHING

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from tabulate import tabulate
```

```python
file_name = input('Excel File Location: ').strip('"')
sheetname = input('Sheet Name: ')
c1 = input('Column name for Period: ')
c2 = input('Column name for Actual data: ')

data = pd.read_excel(r'' + file_name, sheetname)
data_set = pd.DataFrame(data, columns=[c1, c2])

data_set[c2] = data_set[c2].values.astype('float32')
data_set_display = data_set.replace({np.nan: ""})
# data_set[c1] = data_set[c1].replace({np.nan: -1}).astype(int)
# data_set[c1] = np.where(data_set[c1].values.astype(int) > 0, data_set[c1].values.astype(int), None)

alpha = float(input('Enter Alpha: '))
beta = float(input('Enter Beta: '))

print(f'Alpha :  {alpha}')
print(f'Beta :   {beta}')
data_set_display
```

```python
dt = data_set[c2]
lvl = dt.iloc[0]
trnd = 0

level = [lvl]
trend = [trnd]
forecast = [None, lvl + trnd]

for i in range(1, len(data_set)):
    if data_set[c1].iloc[i] == -1:  # Check if the value was originally NaN
        lvl = np.nan
        trnd = np.nan
        frcst = np.nan
    else:
        lvl = alpha * dt.iloc[i] + (1 - alpha) * (level[i-1] + trend[i-1])
        trnd = beta * (lvl - level[i-1]) + (1 - beta) * trend[i-1]
        frcst = lvl + trnd

    level.append(lvl)
    trend.append(trnd)
    forecast.append(frcst)

    # print('{:.2f}'.format(lvl))
    # print('{:.2f}'.format(trnd))
    # print('{:.2f}'.format(frcst))

print(level)
print(trend)
print(forecast)
```

```python
min_length = min(len(dt), len(level), len(trend), len(forecast))
data_set['Level'] = level[:min_length]
data_set['Trend'] = trend[:min_length]
data_set['Forecast'] = forecast[:min_length]

data_set = data_set.round(2)

# Convert -1 back to NaN
data_set[c1] = data_set[c1].replace({-1: np.nan})

header = [c1, c2, "Level", "Trend", "Forecast"]
print(tabulate(data_set.replace({np.nan: ""}), header, tablefmt="grid"))

plt.figure(figsize=(12, 6))

# plt.ylim((min(10), max(50)))

# x_values = data_set[c1].tolist()
# x_values.append(x_values[-1] + 1)
plt.xlim(min(dt), max(dt))

plt.plot(dt, data_set[c2], label='Actual Data', marker='o')
plt.plot(dt, data_set['Forecast'], label='Forecast', linestyle='--', marker='o')

plt.title('Double Exponential Smoothing')
plt.xlabel(c1)
plt.ylabel('Value')
plt.legend()
plt.show

# data_set[c2].plot()
# data_set['Forecast'].plot()
```
